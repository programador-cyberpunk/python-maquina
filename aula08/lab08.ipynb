```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (confusion_matrix, accuracy_score, precision_score,
recall_score, f1_score, fbeta_score,
roc_curve, auc, precision_recall_curve)
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import NearMiss

sns.set_theme(style="whitegrid")
np.random.seed(42)
```

```python
# o nome ja diz ne porra
X_cru ,y_cru = make_classification(n_samples=50000, n_features=20, n_informative=8,
                                   n_redundant=2, weights=[0.9983, 0.0017], 
                                   n_clusters_per_class=1, random_state=42)
# agora vamo organizar essa porra num dataframe
feature_cols = [f"V{1}" for 1 in range (1, 11)]
df = pd.DataFrame(X_cru, columns=feature_cols)
df['Class'] = y_cru

#verifica a contagem absoluta e percentual das classes
class_coluns = df['Class'].value_counts()
class_pcts = df['Class'].value_counts(normalize=True) *100 #normalizando a classe

print("Distribuição amostral de dados bruta: ")
for cls, count in class_coluns.items():
    print(f"Classe {cls}: {count:5d} transações ({class_pcts[cls]: .3f}%)")
```
