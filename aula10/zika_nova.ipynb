```python
import sys
import json
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

def main():
    input_data = sys.stdin.read()
    if not input_data.strip():
        return
    data = json.loads(input_data)

    # Extração dos dados do JSON de entrada
    train_texts = data["train_texts"]
    train_labels = data["train_labels"]
    test_texts = data["test_texts"]
    alpha = float(data.get("alpha", 1.0)) # Padrão alpha=1.0 caso não seja especificado

    # Instanciação do Pipeline combinando CountVectorizer e MultinomialNB (com o alpha especificado)
    pipeline = Pipeline([
        ('vectorizer', CountVectorizer()),
        ('classifier', MultinomialNB(alpha=alpha))
    ])

    # Ajuste (treinamento) do modelo com o conjunto de treinamento
    pipeline.fit(train_texts, train_labels)

    # Execução da inferência (predição) sobre o subconjunto de teste
    predictions = pipeline.predict(test_texts)

    # Conversão das predições para uma lista nativa de Python (caso retornem como numpy array)
    predictions_list = predictions.tolist()

    # Exibição no formato solicitado
    print(f"Predictions: {predictions_list}")

if __name__ == "__main__":
    main()
```
