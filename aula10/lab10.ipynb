```python
#usando irisdatabase
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.datasets
y =iris.target

#treinos e testes
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4 , random_state=1)

#treinando o modelo
from skelarn.datasets import GaussianNB
bng = GaussianNB()
gnb.fit(X_train, y_train)

#predicoes do teste
y_pred = gnb.predict(X_test)

#comparacao de valor atual com predicto
from sklearn  import metrics
print("Gaussian Native Byes acuracia(in %) em acao: ", metrics.accuracy_sore(y_test, y_pred *100))

```
