```python
import numpy as numpy
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
#datasets
california = fetch_california_housing(as_frame=True)
df = california.frame

#dando um visu
display(df.head())
X_real = df.drop(columns=['MedHouseVal'])
Y_real = df['MedHouseVal']

print(f"Total de registros (imoveis): {df.shape[0]}")
print(f"Total de registros atributos: {X_real.shape[1]}")
```

```python
import warnings
# Suprimindo alertas de terceiros
warnings.filterwarnings("ignore", category=DeprecationWarning)

import plotly.graph_objects as go
import numpy as np

amostra = df.sample(1000, random_state=42)

# Selecionando amostras e features pro teste
X_amostra = amostra[['MedInc', 'HouseAge']].values
Y_amostra = amostra[['MedHouseVal']].values.reshape(-1, 1)

# Adiciona coluna de 1 pra calcular o viés/intercepto
X_matriz = np.c_[np.ones((X_amostra.shape[0], 1)), X_amostra]

# Treinamento matemático via álgebra linear
pesos_beta = np.linalg.inv(X_matriz.T @ X_matriz) @ X_matriz.T @ Y_amostra
vies = pesos_beta[0][0]
peso_renda = pesos_beta[1][0]
peso_idade = pesos_beta[2][0]

print("--- PESOS CALCULADOS MANUALMENTE VIA ÁLGEBRA LINEAR ---")
print(f"Intercepto (PREÇO BASE): {vies: .4f}")
print(f"Peso da renda: {peso_renda: .4f}")
print(f"Peso da idade: {peso_idade: .4f}")

# Criando a nuvem de pontos (casas reais)
scatter = go.Scatter3d(
    x=X_amostra[:, 0],
    y=X_amostra[:, 1], # Corrigido: antes estava y=Y_amostra[:, 1] o que gerava erro de dimensão
    z=Y_amostra.flatten(),
    mode='markers',
    marker=dict(size=4, color='blue', opacity=0.5),
    name='Casas Reais'
)

# Criando a malha para o plano matemático
x_surf = np.linspace(X_amostra[:, 0].min(), X_amostra[:, 0].max(), 20)
y_surf = np.linspace(X_amostra[:, 1].min(), X_amostra[:, 1].max(), 20)
x_surf, y_surf = np.meshgrid(x_surf, y_surf)

# Calculando o Z do plano com base nos pesos calculados
z_surf = vies + (peso_renda * x_surf) + (peso_idade * y_surf)

# Criando a superfície do plano matemático
plano = go.Surface(
    x=x_surf,
    y=y_surf,
    z=z_surf,
    colorscale='Reds',
    opacity=0.6,
    name='Modelo (Plano Matemático)'
)

# Montando e renderizando o gráfico
fig = go.Figure(data=[scatter, plano])
fig.update_layout(
    title='Matemática agindo: gire o gráfico aí',
    scene=dict(
        xaxis_title='Renda Média',
        yaxis_title='Idade da Casa',
        zaxis_title='Preço (100k Dólares)'
    ),
    width=900,
    height=700,
    margin=dict(l=0, r=0, b=0, t=40) # Corrigido: antes estava 1=0 (número 1 em vez da letra l)
)

fig.show()
```
