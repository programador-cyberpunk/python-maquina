```python
#exemplo pratico
import matplotlib.pyplot as plt
import numpy as np
import pndas as pd
from sklearn.linear_model import LinearRegression

#dados de um arquivos csv que mostra a relaçao entre o PIB per capita e a qualidade de vida
lifestat = pd.read_csv("lifesat.csv")
X = lifestat[["GDP per capita (Dolleta)"]].values
Y = lifestat[["Satisfação da vida"]].values

#vendo os dados  agora, num grafico de dispersão pra entender melhor
lifestat.plot(kind='scatter', grid=True, x="GDP per capita (Dolleta)", 
                            y="Satisfação da vida")
plt.axis([23_500, 62_500, 4, 9])
plt.show()

# agora escolhe o treinamento do modelo, instanciando o modelo de regressao linear
model = LinearRegression()
model.fit(X, Y)

#prevendo igual a mãe dinah
X_novo = [[37_655.2]]
print(model.predict(X_novo))
```
