```python
import pandas as pd
import numpy as np

#criando uma series vazia, que vai mudar de tipo de  dtype para object,sei la pq
vazio = pd.Series(dtype='object')
print("Nao tem porra nenhuma aqui: ")
print(vazio)

print("-" * 45)

#agora, criando uma Series a partir de um numpy, cada elemento da series vai receber um indice
#numerico padrao ,começando pelo 0
dados_array = np.array(['S', 'a', 'm', 'a', 'e', 'l'])
serie_letras = pd.Series(dados_array)
print("\n Uma serie criada a partir dum array do capeta: ")
print(serie_letras)
```

```python
#dataframe vazio
df_vazio = pd.DataFrame()
print("Ta vazio essa porra: ")
print(df_vazio)
print("-" * 45)

#criando a partir de uma lista
lista_simples = ['vai', 'se', 'fude', 'seu', 'viado']
df_lista = pd.DataFrame(lista_simples)
print("\n Dataframe criado a partir dessa lista: ")
print(df_lista)
print("-" * 45)

dados_dict = { 'Nome': ['Samael', 'Lúcifer', 'Belzebu', 'Asmodeus'],
              'Idade': [666,777,888,999],
              'Cidade': ['Quinto ciclo', 'Setimo ciclo', 'Segundo Ciclo', 'Terceiro ciclo'],
              'Profissao':['FDP', 'FDP', 'FDP', 'FDP']}

df = pd.DataFrame(dados_dict)
print("\n Data frame criado a partir do inferno: ")
print(df)
```

```python
# selecionando colunas
print("Dataframe terra 1: ")
print(df)
print("-" * 45)

#selecionar uma só coluna
series_nomes = df['Nome']
print("\n Selecionando a coluna nome: ")
print(type(series_nomes))
print(series_nomes)
print("-" * 45)

#AGORA SÃO MULTIPLAS COLUNBAS
sub_df = df[['Nome', 'Cidade']]
print("\n Selecionando multiplas colunas: ")
print(type(sub_df))
print(sub_df)
```
