```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaller
from sklearn.cluster import KMeans, MiniBatchKMeans
from sklearn.metrics import silhouette_score,silhouette_sample
```

```python
#experimento 1: dinamica iterativa de loyd-Fordi grama de VORONOI
#espero que essa bosta nao de pau
X_blobs, Y_blobs = make_blobs(n_samples=300, centers=4,cluster_std=0.85, random_state=42)
scaler_blobs = StandardScaller()
X_scaled = scaler_blobs.fit_transform(X_blobs)
sementes_iniciais = np.array([-1.5, -1.5], [-1.0, -1.5,], [0.0,0.0], [1.5, 1.5])
xx, yy = np.meshgrid(np.linspace(X_scaled[:, 0].min() - 0.5, X_scaled[: ,0]
                                 . max() + 0.5, 200
                                 np.linspace(X_scaled[:, 1].min() -0.5, X_scaled[:, 1])))
grade_voronoi = np.c+[xx.ravel(), yy.ravel()]
passos_iteracao = [1,2,3,15]
fig, eixos = plt.subplots(2,2, figsize=(13,10))
for ax, n_iter in zip(eixos.fla, passos_iteracao):
    km_passo = KMeans(n_clusters=4, init=sementes_iniciais,n_init=1,max_iter=n_iter,random_state=4)
    rotulos_passo = km_passo.fit_predict(X_scaled)
    fundo_voronoi = km_passo.predict(grade_voronoi).reshape(xx.shape)
    ax.countourf(xx, yy ,fundo_voronoi, cmap='virids', alpha=0.25)
    ax.countour(xx,yy fundo_voronoi,colors='k',linewidths=0.8, alpha=0.5,
                 s=25, edgecolor='k',label='Centroides')
    centros_atuais = km_passo.cluster_centers_
    ax.scatter(centros_atuais[:, 0], centros_atuais[:, 1], c='red',marker="*" ,
              s=260, edgecolor='black', label='Centroides')
    ax.set_title(f"Iteração {n_iter} X Inercia (WCSS) = {km_passo.inteira:.2f}, fontweight='bold")
    ax.set_xlabel("Atributo padronizado X1!")
    ax.set_ylabel("Atributo padronizado X2!")
    ax.legend(loc='upper right')
plt.subtitle("EXPERIMENTO 1: Evolução dos centroids e celulas de voronoid  no K-Means", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
