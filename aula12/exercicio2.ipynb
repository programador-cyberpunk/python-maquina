```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
#exercicio 2, risco de falha industrial(n=10)
#atributo temperatura alta =1, normal=0 , alvo falha? sim=1, nao=0

X1 = np.array([[1], [1], [1], [1], [0], [0], [0], [0]])
Y1 = np.array([1, 1, 1, 0, 1, 0, 0, 0])

clf2 = DecisionTreeClassifier(random_state=0)
clf2.fit(X1, Y1)

plt.figure(figsize=(10, 6))
plot_tree(clf2, feature_names=['Temperatura'], 
          class_names=['Não', 'Sim'],
          filled=True,
          rounded=True)
plt.title("Falha industrial")
plt.show()
```
