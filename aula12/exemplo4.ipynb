```python
from sklearn.tree import DecisionTreeCLassifier
from sklearn.datasets import load_iris

def rank_importancia_atributos(X, y, feature_names, max_depth=4):
    #vamo cria essa porra de arvore filha da puta de decisao
    clf = DecisionTreeClassifier(max_depth = max_depth, random_state=42)
    #fica vendo, essa porra vai dar erro de novo e eu quero que se foda
    clf.fit(X, y)
    importances = clf.feature_importances_
    ranked = sorted(zip(feature_names, importances), key=lambda x: x[1], reverse=True)
    return [(name, round(float(imp), 4)) for name, imp in ranked]
iris = load_iris()
ranking = rank_importancia_atributos(iris.data, iris.target, iris.feature_names)
print("Rakngin de importancia dessas porra: ")
for feat, imp in rakning:
    print(f" - {feat:20s}: {imp}")
```
