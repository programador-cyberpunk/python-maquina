```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python

#exercicio 1, o famoso golpe do crediario, atributo renda
#o alvo é o emprestimo aprovado
X1 = np.array([[1], [1], [1], [1], [0], [0], [0], [0]])
Y1 = np.array([1, 1, 1, 0, 0, 0, 0, 0])

clf1 = DecisionTreeClassifier(criterion='entropy', random_state=42)
clf1.fit(X1, Y1)

plt.figure(figsize=(5, 3))
plot_tree(clf1, feature_names=['Renda'], 
          class_names=['Não', 'Sim'], 
          filled=True,
          rounded=True)
plt.title("Risco de Credito")
plt.show()
```
