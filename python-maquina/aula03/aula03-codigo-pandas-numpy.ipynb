```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#graficos
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 6)
pd.set_option('display.float_format', lambda x: '%.4f' % x)

print("====== TUDO NOS TRINKS ======")

print(f"NumPy Version: {np.__version__}")
print(f"Pandas Version: {np.__version__}")
```

```python
#dataset
np.random.seed(42)
n = 300
idade = np.random.normal(loc=38, scale=12, size=n)

#variavel continua dependente
renda = (idade *150) +np.random.normal(loc=3500, scale=1800, size=n)

 #Variavel continua dependente
score = 850 - (idade * 4) + np.random.normal(loc=0, scale=30,size=n)

#vaariavel categorica discreta
escolaridade = np.random.choice(['Ensino Medio', 'Graduação', 'Pós-Graduação','Mestrado;Doutorado'],
                                size = n, p=[0.25,0.45,0.20, 0.10])
#agora vem o Pandas
df = pd.DataFrame({
    'Idade': idade,
    'Renda': renda,
    'Escolaridade': escolaridade
})
print(" --- AGORA VEM AS PRIMEIRAS LINHAS DE DADOS ---")
display(df.head())
```

```python
# extração
print(" === MEDIDas DE TENDENCIA CENTRAL === ")
print(f"Media da renda mensal: R$ {df['Renda'].mean():.2f}")
print(f"Mediana da renda mensal: R$ {df['Renda'].median():.2f}")

#moda extraida usando o mode.()[0], pq podem ter valores empatados
moda_escolar = df['Escolaridade'].mode()[0]
print(f"Moda da escolaridade: '{moda_escolar}'")
print("-> Uso pratico (engenharia de dados): Se o criente não preencheu  o  campo idade ", "podemos utilizar a Moda para imputar este valor nulo preenchimento␣
↪seguro.")
print(" === MEDIDAS DE DISPERSÃO E ESPALHAMENTO ===")
var_score = df['Score'].var(ddof=1)
std_score = df['Score'].std(ddof=1)
print(f"Variancia do score(s²) : {var_score:.2f} pontos²")
print(f"Desvio padrao 06 do score(s²) : {std_score:.2f} pontos de afastamento")
```

```python
#Associação: covariancia e correlação de pearson

df_num = df[['Idade', 'Renda', 'Score']]
print('=== MATRIX COVARIANCIA BRUTA ===')
display(df_num.cov())

print("=== MATRIX DE CORRELAÇAO NORMALIZADA DE PEARSON")
display(df_num.corr())

#dando um visu na associação
plt.figre(figsize=(9, 6))
sns.pairplot(df_num, kind='reg', diag_kind='kde',
            plot_kws={{'line_kws'}: {'color': 'crimson', 'lw': 2},
            'sccatter_kws': {'alpha': 0.5}})
plt.show()
```

```python
#COVARIANCIA MATRICIAL DO SCRATCH EM NUMPY

#primeiro, isolar os dados com o Numpy
X = df_num.values
n_linhas, p_atributos = X.shape

#vetorizando
vetor_medias = np.mean(X, axis=0)
#centralizando
X_centralizado = X -vetor_medias
#tensiorial escalar para a matriz
matriz_sigma_manual = (X_centralizado.T @ X_centralizado) / (n_linhas -1)
df_sigma = pd.DataFrame(matriz_sigma_manual, index=df_num.colmuns, columns=df_num.columns)
print("=== MATRIX DE COVARIANCIA OBTIDA MANUALMENTE VIA ALGORITMO ===").display(df_sigma)

#tirando a prova de assertividade computacional
assert np.allclose(matriz_sigma_manual, np.cov(X, rowvar = False)), "Divergencia matematica flagrada"
print("Validação do celero matricial: Operações de matriz manual casão certinho com os calculos do Numpy e Pandas")
```
