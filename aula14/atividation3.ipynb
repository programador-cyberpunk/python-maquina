```python
import numpy as np
```

```python
def predizer_meta_learner_stacking(P_camada0, w0, w_meta):
    P = np.asarray(P_camada0, dtype=float)
    w_meta = np.asarray(w_meta, dtype=float)
    
    # PASSO 1: Calcule o vetor de escores lineares z_meta = w0 + produto escalar entre P e w_meta
    z_meta = w0 + np.dot(P, w_meta)
    
    # PASSO 2: Gere o vetor de classes inteiras (1 se z_meta >= 0.0, senão 0)
    classes_preditas = (z_meta >= 0.0).astype(int)
    
    return np.round(z_meta, 3).tolist(), classes_preditas.tolist()
```
